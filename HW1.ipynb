### Influenza Vaccine Doses Dataset

**General Information:**  
This dataset contains information about influenza vaccine doses distributed in the United States during different influenza seasons.

**Features/Columns:**

1. **Influenza_Season**
   - Identifies the influenza season associated with the vaccine distribution data.

2. **Start_Date**
   - Shows the beginning date of the reporting period.

3. **End_Date**
   - Shows the ending date of the reporting period.

4. **Reporting_TimeFrame**
   - Describes the time period used for reporting the vaccine distribution data.

5. **Week_Sort_Order**
   - Shows the numerical order of the reporting week.

6. **Current_Season_Week_Ending_Label**
   - Shows the ending date of the current reporting week.

7. **Cumulative_Flu_Doses_Distributed**
   - Shows the cumulative number of influenza vaccine doses distributed.

8. **Current_Through**

### U.S. State Life Expectancy Dataset
